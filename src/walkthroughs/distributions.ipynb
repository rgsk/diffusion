```python
import torch
```

```python
N = 100000
```

```python
def pn(n):
    print(f"{n:.3f}")
```

### Continuous


```python
a = 9
b = 90
x = torch.rand(N) * (b - a) + a
pn(x.mean().item())
```

```python
pn((a + b) / 2)
```

```python
pn(x.var(unbiased=False).item())
```

```python
pn((b - a) ** 2 / 12)
```

### Discrete


```python
a = 1
b = 6
x = torch.randint(a, b + 1, (N,), dtype=torch.float)
pn(x.mean().item())
```

```python
pn((a + b) / 2)
```

```python
pn(x.var(unbiased=False).item())
```

```python
pn(((b - a + 1) ** 2 - 1) / 12)
```
